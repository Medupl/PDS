## <span style="color: #1A5FA8;">1. O Processo de Amostragem</span>

A **amostragem periódica** é o primeiro passo na conversão analógico-digital (ADC).
Dado um sinal contínuo $x_c(t)$, a amostragem extrai valores em instantes uniformemente espaçados:

$$\boxed{x(n) = x_c(n\,T_s), \quad n = 0, \pm1, \pm2, \ldots}$$

onde $T_s = 1/F_s$ é o período de amostragem e $F_s$ é a frequência de amostragem (Hz).

### Modelo matemático — trem de impulsos

O processo é modelado pela multiplicação do sinal contínuo por um trem de impulsos periódico**:

$$s(t) = \sum_{n=-\infty}^{+\infty} \delta(t - nT_s)$$

$$x_s(t) = x_c(t)\cdot s(t) = \sum_{n=-\infty}^{+\infty} x_c(nT_s)\,\delta(t - nT_s)$$

<div style="text-align: center;">
  
  <p style="color: gray; font-size: 0.9em;">Figura 1 — Diagrama do processo de amostragem</p>
</div>

### Relação entre frequências

| Grandeza | Símbolo | Unidade |
|---|---|---|
| Frequência de amostragem | $F_s$ | Hz |
| Período de amostragem | $T_s = 1/F_s$ | s |
| Frequência digital (normalizada) | $\omega = \Omega\,T_s = 2\pi F/F_s$ | rad/amostra |
| Frequência de Nyquist | $F_N = F_s/2$ | Hz |


>Ponto importante:</strong> a amostragem discretiza o tempo, mas não a amplitude. A quantização (conversão da amplitude para representação binária) é um processo separado e independente da amostragem.

## <span style="color: #1A5FA8;">2. Análise Espectral da Amostragem</span>

### Espectro do trem de impulsos

O trem de impulsos $s(t)$ com período $T_s$ tem transformada de Fourier:

$$S(j\Omega) = \frac{2\pi}{T_s}\sum_{k=-\infty}^{+\infty}\delta(\Omega - k\,\Omega_s), \qquad \Omega_s = \frac{2\pi}{T_s}$$

### Espectro do sinal amostrado

Como a multiplicação no tempo corresponde à convolução no domínio da frequência:

$$X_s(j\Omega) = \frac{1}{2\pi}\,X_c(j\Omega) * S(j\Omega) = \frac{1}{T_s}\sum_{k=-\infty}^{+\infty} X_c\!\left(j(\Omega - k\,\Omega_s)\right)$$

O espectro do sinal amostrado é uma soma de réplicas deslocadas do espectro original, repetidas a cada $\Omega_s = 2\pi F_s$.

<div style="text-align: center;">
  
  <p style="color: gray; font-size: 0.9em;">Figura 2 — Espectros dos sinais: caso sem sobreposição ($\Omega_s > 2\Omega_N$)</p>
</div>

>Condição de recuperação:</strong> quando as réplicas não se sobrepõem, o sinal original pode ser recuperado por um filtro passa-baixa ideal com frequência de corte $\Omega_c = \Omega_s/2$. Se houver sobreposição, o aliasing torna a recuperação impossível.



---

## <span style="color: #1A5FA8;">3. Teorema da Amostragem (Nyquist-Shannon)</span>

### Enunciado

> Seja $x_c(t)$ um sinal de banda limitada com $X_c(j\Omega) = 0$ para $|\Omega| > \Omega_N$.
> Então $x_c(t)$ é unicamente determinado por suas amostras $x_c(nT_s)$ se e somente se:
>
> $$\boxed{F_s > 2\,F_N}$$
>
> onde $F_N = \Omega_N / 2\pi$ é a frequência máxima (frequência de Nyquist) do sinal.

A reconstrução exata é feita por um filtro passa-baixa ideal com ganho $T_s$ e frequência de corte $F_c$ tal que $F_N < F_c < F_s - F_N$ (tipicamente $F_c = F_s/2$).

### Conceitos derivados

| Conceito | Definição |
|---|---|
| **Frequência de Nyquist** $F_N$ | Maior frequência presente no sinal ($F_{max}$) |
| **Taxa de Nyquist** | Menor $F_s$ sem aliasing: $F_s^{min} = 2F_N$ |
| **Intervalo de Nyquist** | Período máximo de amostragem: $T_s^{max} = 1/(2F_N)$ |

### Casos práticos

| Aplicação | $F_{max}$ | $F_s$ mínimo | Padrão usado |
|---|---|---|---|
| Voz telefônica | 4 kHz | 8 kHz | 8 kHz (G.711) |
| VoIP / Wideband | 8 kHz | 16 kHz | 16 kHz |
| Áudio CD | 20 kHz | 40 kHz | **44,1 kHz** |
| Áudio profissional | 24 kHz | 48 kHz | 48/96 kHz |

>Por que o CD usa 44,1 kHz e não 40 kHz?</strong><br>
A margem extra (10,25%) é reservada para a banda de transição do filtro anti-aliasing real, que não possui queda abrupta — precisa de espaço entre a banda de passagem (20 kHz) e a banda de rejeição (início do dobramento).

</div>

## <span style="color: #1A5FA8;">4. Aliasing e Mapeamento de Frequências</span>

### O que é aliasing?

Quando $F_s < 2F_N$, as réplicas espectrais se sobrepõem,  fenômeno chamado de *aliasing*.
Uma componente de frequência $F > F_s/2$ não desaparece; ela se "disfarça" como uma frequência mais baixa.

>Consequência irreversível:</strong> uma vez que o aliasing ocorre durante a amostragem, não é possível desfazê-lo no processamento digital posterior. A única proteção eficaz é o filtro anti-aliasing aplicado <em>antes</em> do amostrador.

### Mapeamento de frequências

Uma componente analógica de frequência $F$ é mapeada para a frequência digital:

$$\omega = \frac{2\pi F}{F_s}$$

Se $F > F_s/2$, a frequência observada no espectro digital é:

$$\boxed{F_{alias} = \left|\,F - \text{round}\!\left(\frac{F}{F_s}\right)\cdot F_s\,\right|}$$

### Gráfico de dobramento (*folding diagram*)

O gráfico de dobramento mostra como cada frequência real $F$ é percebida no domínio digital.
Trata-se de uma função "dente de serra" triangular com período $F_s$:

<div style="text-align: center;">
  
  <p style="color: gray; font-size: 0.9em;">Figura 4 — Gráfico de dobramento: frequência real vs. frequência percebida</p>
</div>

### Exemplo numérico

Com $F_s = 40$ Hz e uma senóide de $F = 35$ Hz:

$$F_{alias} = |35 - 1 \times 40| = 5 \text{ Hz}$$

A senóide de 35 Hz soa como uma senóide de 5 Hz — indistinguível pelas amostras!

| $F_0$ (Hz) | $F_s = 40$ Hz | $F_{alias}$ (Hz) | Status |
|:---:|:---:|:---:|:---:|
| 5 | 40 | 5 | ✅ Sem aliasing |
| 19 | 40 | 19 | ✅ Sem aliasing |
| 21 | 40 | 19 | ⚠️ Aliasing! |
| 35 | 40 | 5 | ⚠️ Aliasing! |
| 60 | 40 | 20 | ⚠️ Aliasing! |

### Exemplos cotidianos de aliasing

| Situação | O que acontece |
|---|---|
| Roda de carro em câmera lenta | Roda parece girar ao contrário |
| Hélice de avião em vídeo | Hélice parece parada ou lenta |
| Padrão fino em imagem (moiré) | Surge um padrão falso de baixa frequência |
| Áudio gravado com $F_s$ baixo | Tom alto soa como tom mais baixo |

## <span style="color: #1A5FA8;">5. Reconstrução por Interpolação Sinc </span>

### Fórmula da reconstrução ideal

Satisfeita a condição de Nyquist, o sinal original pode ser perfeitamente reconstruído
a partir das amostras pela fórmula de interpolação sinc:

$$\boxed{x_r(t) = \sum_{n=-\infty}^{+\infty} x[n] \cdot \operatorname{sinc}\!\left(\frac{t - nT_s}{T_s}\right)}$$

onde a função sinc normalizada é:

$$\operatorname{sinc}(\alpha) = \frac{\operatorname{sen}(\pi \alpha)}{\pi \alpha}, \qquad \operatorname{sinc}(0) = 1$$

### Por que funciona?

Cada parcela $x[n]\cdot\operatorname{sinc}\!\left(\frac{t-nT_s}{T_s}\right)$ é uma função sinc centrada em $t = nT_s$ e escalada por $x[n]$. A propriedade fundamental da sinc garante:

$$x[n]\cdot\operatorname{sinc}(k-n) = \begin{cases} x[n] & k = n \\\\ 0 & k \neq n \end{cases}$$

Cada sinc vale exatamente o valor da amostra no seu instante e zero nos demais — eliminando interferência entre termos.

<div style="text-align: center;">
  
  <p style="color: gray; font-size: 0.9em;">Figura 5 — Superposição de funções sinc centradas nos instantes de amostragem</p>
</div>

### Equivalência com filtro passa-baixa ideal

No domínio da frequência, a reconstrução sinc equivale a um filtro passa-baixa ideal:

$$H_r(j\Omega) = \begin{cases} T_s & |\Omega| \leq \pi/T_s \\\\ 0 & |\Omega| > \pi/T_s \end{cases}$$

cuja resposta ao impulso é exatamente a função sinc: $h_r(t) = \operatorname{sinc}(t/T_s)$.

### Limitações práticas

| Problema | Causa | Solução prática |
|---|---|---|
| Filtro ideal irrealizável | Sinc tem suporte infinito | Janelamento (Hann, Kaiser, etc.) |
| Truncamento temporal | Sinal sempre finito | Aumentar $N$; aceitar erro de borda |
| Aliasing residual | $F_s < 2F_{max}$ | Filtro anti-aliasing antes do ADC |
| Retentor de ordem zero | DAC real usa ZOH | Compensação espectral pós-reconstrução |



---

## <span style="color: #1A5FA8;">6. Condições para Reconstrução Perfeita</span>

Para que $x_r(t) = x_c(t)$ exatamente, todas as quatro condições abaixo devem ser satisfeitas simultaneamente:


1. **Banda limitada:** $X_c(j\Omega) = 0$ para $|\Omega| > \pi F_s$
2. **Nyquist satisfeito:** $F_s > 2F_{max}$
3. **Filtro ideal:** $H_r(j\Omega) = T_s$ para $|\Omega| \leq \pi/T_s$, zero caso contrário
4. **Sem quantização:** $B \to \infty$ (ou erro de quantização desprezível)


Na prática, nenhuma dessas condições é satisfeita exatamente — o que torna o projeto de sistemas ADC/DAC reais um compromisso entre qualidade, custo e complexidade computacional.

## <span style="color: #1A5FA8;">7. Simulador Interativo</span>

A célula seguinte lança o simulador com slider dinâmico. Ao mover o controle de $F_0$, todos os 10 gráficos são atualizados em tempo real:

| Gráfico | O que mostra |
|---|---|
| 1. Sinal Contínuo | $x_c(t) = \sin(2\pi F_0 t)$ |
| 2. Trem de Impulsos | $s(t) = \sum \delta(t - nT_s)$ |
| 3. Amostras $x[n]$ | Valores discretos nos instantes de amostragem |
| 4. Sincs (zoom) | Primeiras 10 funções sinc individuais |
| 5. Reconstrução vs Amostras | $x_r(t)$ sobreposto às amostras |
| 6. Erro | $e(t) = x_c(t) - x_r(t)$ |
| 7. FFT Original | $|X_c(f)|$ |
| 8. FFT Reconstruída | $|X_r(f)|$ |
| 9. Sobreposição espectral | Original vs reconstruído |
| 10. Gráfico de dobramento | $F_{alias}$ vs $F_0$ real |

> Roteiro de experimentos sugerido:
>
>1. **$F_0 = 5$ Hz** → reconstrução perfeita, erro ≈ 0. Observe o gráfico 10: ponto na diagonal.
>2. **$F_0 = 19$ Hz** → ainda sem aliasing, mas próximo do limite. Erro nas bordas aumenta.
>3. **$F_0 = 20$ Hz** → exatamente no Nyquist. O que acontece? (caso limite!)
>4. **$F_0 = 21$ Hz** → aliasing ativo. Veja o ponto "dobrar" no gráfico 10.
>5. **$F_0 = 35$ Hz** → confirme que $F_{alias} = |35 - 40| = 5$ Hz — igual ao caso 1!
>6. **$F_0 = 60$ Hz** → caso extremo. Observe o espectro no gráfico 9.


```python
import numpy as np
import matplotlib
import matplotlib.collections
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.widgets import Slider

# Tenta forçar o backend interativo para abrir janela externa
try:
    matplotlib.use('TkAgg')
except Exception:
    pass

def gerar_simulador():
    # ─── Parâmetros de Configuração ───────────────────────────────────────
    F0_init = 5.0    # Frequência inicial da senoide (Hz)
    Fs      = 40.0   # Taxa de amostragem (Hz)  — Nyquist = 20 Hz
    T_dur   = 1.0    # Duração total (s)
    A       = 1.0    # Amplitude
    Ts      = 1.0 / Fs

    # ─── Paleta de Cores ──────────────────────────────────────────────────
    C_CONT = "#1A5FA8"   # azul  — sinal contínuo
    C_SAMP = "#C0392B"   # vermelho — amostras / trem de impulsos
    C_REC  = "#117A65"   # verde — sinal reconstruído
    C_ERRO = "#8E44AD"   # roxo  — erro de reconstrução
    C_NYQ  = "#E67E22"   # laranja — frequência de Nyquist

    # ─── Grades de Dados ──────────────────────────────────────────────────
    t_cont    = np.linspace(0, T_dur, 1000)
    t_samp    = np.arange(0, T_dur, Ts)
    N_fft     = 2048
    freqs_fft = np.fft.rfftfreq(N_fft, d=t_cont[1] - t_cont[0])

    # ─── Figura e Layout ──────────────────────────────────────────────────
    fig = plt.figure(
        figsize=(15, 12),
        num="Simulador de Amostragem e Reconstrução Sinc"
    )
    gs = gridspec.GridSpec(4, 3, figure=fig, hspace=0.6, wspace=0.3)
    plt.subplots_adjust(bottom=0.08, top=0.92, left=0.07, right=0.97)

    # 9 subplots principais + gráfico de dobramento (linha inteira)
    axes = [fig.add_subplot(gs[i, j]) for i in range(3) for j in range(3)]
    ax1, ax2, ax3, ax4, ax5, ax6, ax7, ax8, ax9 = axes
    ax_fold = fig.add_subplot(gs[3, :])

    # ─── Objetos Gráficos (criados uma única vez) ─────────────────────────

    # Gráfico 1 — Sinal contínuo
    line1, = ax1.plot(t_cont, np.zeros_like(t_cont),
                      color=C_CONT, lw=1.5, label="x_c(t)")

    # Gráfico 3 — Amostras (stem)
    stem3 = ax3.stem(t_samp, np.zeros_like(t_samp),
                     linefmt=C_SAMP, basefmt="gray",
                     markerfmt='ro', label="x[n]")

    # Gráfico 4 — Primeiras 10 funções sinc individuais (zoom)
    n_sincs_zoom = 10
    colors_sinc  = plt.cm.Dark2(np.linspace(0, 1, n_sincs_zoom))
    sinc_lines   = [
        ax4.plot(t_cont, np.zeros_like(t_cont), '--',
                 color=colors_sinc[i], lw=1.2)[0]
        for i in range(n_sincs_zoom)
    ]

    # Gráfico 5 — Reconstrução + amostras sobrepostas
    line5, = ax5.plot(t_cont, np.zeros_like(t_cont),
                      color=C_REC, lw=2.5, zorder=3, label="x_r(t)")
    scat5 = ax5.scatter(t_samp, np.zeros_like(t_samp),
                        color='red', s=30, zorder=10,
                        edgecolors='black', label="Amostras")

    # Gráficos 7, 8, 9 — Espectros
    line7,  = ax7.plot(freqs_fft, np.zeros_like(freqs_fft),
                       color=C_CONT, label="|X_c(f)|")
    line8,  = ax8.plot(freqs_fft, np.zeros_like(freqs_fft),
                       color=C_REC,  label="|X_r(f)|")
    line9a, = ax9.plot(freqs_fft, np.zeros_like(freqs_fft),
                       color=C_CONT, alpha=0.3, label="Orig.")
    line9b, = ax9.plot(freqs_fft, np.zeros_like(freqs_fft),
                       '--', color=C_REC, label="Recon.")

    # Gráfico 10 — Dobramento teórico
    f_real_axis   = np.linspace(0, 100, 1000)
    f_nyq         = Fs / 2
    f_fold_theory = np.abs((f_real_axis + f_nyq) % Fs - f_nyq)
    ax_fold.plot(f_real_axis, f_fold_theory,
                 color='gray', alpha=0.3, label="Teoria (gráfico de dobramento)")
    dot_fold, = ax_fold.plot([], [], 'ro', ms=10,
                              label="Estado Atual", zorder=5)

    # ─── Configurações Estéticas Fixas ────────────────────────────────────
    titles = [
        "1. Sinal Contínuo  x_c(t)",
        "2. Trem de Impulsos  s(t)",
        "3. Amostras  x[n]",
        "4. Sincs Individuais (zoom)",
        "5. Reconstrução  x_r(t)  vs  Amostras",
        "6. Erro  e(t) = x_c(t) − x_r(t)",
        "7. Espectro Original  |X_c(f)|",
        "8. Espectro Reconstruído  |X_r(f)|",
        "9. Sobreposição Espectral",
    ]

    for i, ax in enumerate(axes):
        ax.set_title(titles[i], fontsize=9, fontweight='bold')
        ax.grid(True, alpha=0.2, ls=':')
        if ax in [ax1, ax3, ax5, ax7, ax8, ax9]:
            ax.legend(loc="upper right", fontsize=7)
        if ax in [ax1, ax3, ax5, ax6]:
            ax.set_ylim([-1.5, 1.5])

    for ax in [ax7, ax8, ax9]:
        ax.set_xlim([0, Fs])
        ax.set_ylim([0, 1.2])
        ax.axvline(f_nyq, color=C_NYQ, ls="--",
                   label=f"Nyquist = {f_nyq} Hz")

    ax4.set_xlim([0, 0.25])
    ax4.set_ylim([-0.8, 1.2])
    ax4.set_title("4. Sincs Individuais (zoom: 0–250 ms)", fontsize=9, fontweight='bold')

    ax_fold.set_title(
        "10. Gráfico de Dobramento — Frequência Real F₀ vs Frequência Percebida F_alias",
        fontweight='bold', fontsize=10
    )
    ax_fold.set_xlabel("Frequência real F₀ (Hz)")
    ax_fold.set_ylabel("Frequência percebida F_alias (Hz)")
    ax_fold.axvline(f_nyq, color=C_NYQ, ls="--", alpha=0.5,
                    label=f"Nyquist = {f_nyq} Hz")
    ax_fold.set_xlim([0, 100])
    ax_fold.set_ylim([-2, f_nyq + 10])
    ax_fold.legend(fontsize=8)

    # ─── Função de Atualização Dinâmica ───────────────────────────────────
    def update(val):
        F0 = float(val)
        xc = A * np.sin(2 * np.pi * F0 * t_cont)
        xn = A * np.sin(2 * np.pi * F0 * t_samp)

        # Interpolação Whittaker-Shannon
        xr = np.zeros_like(t_cont)
        for i, (tn, vn) in enumerate(zip(t_samp, xn)):
            sinc_v = vn * np.sinc(Fs * (t_cont - tn))
            xr    += sinc_v
            if i < n_sincs_zoom:
                sinc_lines[i].set_ydata(sinc_v)

        # Limpeza seletiva: remove PolyCollection (fill_between) e textos,
        # mas preserva PathCollection (scatter de amostras)
        for ax in [ax1, ax2, ax5, ax6]:
            for artist in list(ax.collections):
                if isinstance(artist, matplotlib.collections.PolyCollection):
                    artist.remove()
            while ax.texts:
                ax.texts[0].remove()

        # Atualização — Gráfico 1: Sinal Contínuo
        line1.set_ydata(xc)
        ax1.fill_between(t_cont, xc, color=C_CONT, alpha=0.1)

        # Atualização — Gráfico 2: Trem de Impulsos (redesenha setas)
        for ts in t_samp:
            ax2.annotate(
                "", xy=(ts, 1.0), xytext=(ts, 0),
                arrowprops=dict(arrowstyle="-|>", color=C_SAMP, lw=1.5)
            )

        # Atualização — Gráficos 3 e 5: Amostras
        stem3.markerline.set_ydata(xn)
        stem3.stemlines.set_segments(
            [np.array([[ts, 0], [ts, v]]) for ts, v in zip(t_samp, xn)]
        )
        line5.set_ydata(xr)
        ax5.fill_between(t_cont, xc, color=C_CONT, alpha=0.05)
        scat5.set_offsets(np.c_[t_samp, xn])

        # Atualização — Gráfico 6: Erro
        ax6.fill_between(t_cont, xc - xr, color=C_ERRO, alpha=0.4)

        # Atualização — Espectros (7, 8, 9)
        m_orig = np.abs(np.fft.rfft(xc, n=N_fft)) / (N_fft / 4)
        m_rec  = np.abs(np.fft.rfft(xr, n=N_fft)) / (N_fft / 4)
        line7.set_ydata(m_orig)
        line8.set_ydata(m_rec)
        line9a.set_ydata(m_orig)
        line9b.set_ydata(m_rec)

        # Atualização — Gráfico 10: Dobramento
        f_percebida = np.abs((F0 + f_nyq) % Fs - f_nyq)
        dot_fold.set_data([F0], [f_percebida])

        # Título dinâmico com status de aliasing
        status       = "SEM ALIASING ✓" if Fs > 2 * F0 else "ALIASING!"
        color_status = "#0F6E56"        if Fs > 2 * F0 else "#A32D2D"
        f_alias_info = (f"  →  F_alias = {f_percebida:.1f} Hz"
                        if Fs <= 2 * F0 else "")
        fig.suptitle(
            f"F₀ = {F0:.1f} Hz  |  Fs = {Fs} Hz  |  Nyquist = {f_nyq} Hz"
            f"  |  {status}{f_alias_info}",
            fontsize=14, fontweight='bold', color=color_status
        )
        fig.canvas.draw_idle()

    # ─── Slider ───────────────────────────────────────────────────────────
    ax_sl  = plt.axes([0.25, 0.02, 0.5, 0.025], facecolor='#F0F0F0')
    slider = Slider(ax_sl, 'F₀ (Hz)', 1.0, 100.0,
                    valinit=F0_init, valstep=0.5)
    slider.on_changed(update)

    # Primeira renderização e loop
    update(F0_init)
    plt.show()


if __name__ == "__main__":
    gerar_simulador()
```

>Gráficos 1, 3 e 5 (domínio do tempo)</strong><br>
Com $F_0 < F_s/2$: o sinal reconstruído (verde, gráfico 5) sobrepõe quase perfeitamente o original (azul). As amostras vermelhas estão exatamente sobre a senóide. O erro (gráfico 6) é pequeno e concentrado nas bordas temporais.

>Gráfico 6 — Erro de reconstrução</strong><br>
Quando $F_0$ ultrapassa $F_s/2$, o erro explode em toda a janela temporal — não apenas nas bordas. Isso é a assinatura visual do aliasing: a interpolação sinc reconstrói <em>uma senóide diferente</em> da original.

>Gráfico 9 — Sobreposição espectral</strong><br>
Sem aliasing: o pico do espectro reconstruído (verde tracejado) coincide com o original (azul). Com aliasing: o pico migra para $F_{alias}$ — mostrando que o sistema "acredita" estar reconstruindo uma frequência diferente da real.

>Gráfico 10 — Gráfico de dobramento</strong><br>
O ponto vermelho se move sobre a curva triangular. Enquanto está na diagonal (região $F_0 < F_s/2$), $F_{alias} = F_0$ e não há distorção. Quando "dobra" para baixo, o ponto está na região de aliasing — e a frequência percebida é menor que a real.
